```python
%sql

-- =====================================================================
-- Unity Catalog - Cliente AGA
-- Catalog + schemas bronze/silver + tabelas: station, screen,
-- parameter_sample, parameter_result, gw_level, flow_rate
-- =====================================================================

CREATE CATALOG IF NOT EXISTS AGA
    MANAGED LOCATION 's3://databricks-client-dev/aga/';

CREATE SCHEMA IF NOT EXISTS AGA.RAW;
CREATE SCHEMA IF NOT EXISTS AGA.bronze;

CREATE SCHEMA IF NOT EXISTS AGA.silver;


-- =====================================================================
-- BRONZE (tudo STRING, sem PK/NOT NULL)
-- =====================================================================

CREATE TABLE IF NOT EXISTS AGA.bronze.station (
  id                STRING,
  name              STRING,
  x                 STRING,
  y                 STRING,
  elevation         STRING,
  toc               STRING,
  stn_type          STRING,
  depth             STRING,
  purpose           STRING,
  comment           STRING,
  construction_date STRING,
  update_date       STRING,
  data_source       STRING,
  demand_type       STRING,
  stn_description   STRING,
  matrix            STRING,
  bussiness_unit    STRING,
  photo             STRING,
  record_resp       STRING,
  status            STRING,
  watershed         STRING,
  frame_legislation STRING,
  water_body        STRING,
  current_elevation STRING,
  sub_watershed     STRING,
  stn_class         STRING,
  paebm             STRING,
  mecanica_rochas   STRING,
  geotecnia         STRING,
  meio_ambiente     STRING,
  comment_name      STRING,
  start_monitoring  STRING,
  end_monitoring    STRING
);

CREATE TABLE IF NOT EXISTS AGA.bronze.project (
    ID STRING,
    Name STRING,
    plsd STRING,
    Client STRING,
    PDA STRING,
    List STRING,
    Soil STRING,
    Description STRING,
    Path STRING,
    method_name STRING,
    chemical_name STRING
);

CREATE TABLE IF NOT EXISTS AGA.bronze.pstation (
    ID STRING,
    SID STRING,
    x STRING,
    y STRING,
    prj_sys STRING
);

CREATE TABLE IF NOT EXISTS AGA.bronze.screen (
  station         STRING,
  screen_id       STRING,
  from_           STRING,
  to_             STRING,
  diam_           STRING,
  screen_type     STRING,
  slot_number     STRING,
  screen_material STRING,
  comment         STRING,
  data_source     STRING,
  record_resp     STRING
);

CREATE TABLE IF NOT EXISTS AGA.bronze.parameter_sample (
  station              STRING,
  sample_id            STRING,
  sample_date          STRING,
  lab_id               STRING,
  batch_id             STRING,
  quality_code         STRING,
  chain_cust           STRING,
  collection_quarter   STRING,
  comment              STRING,
  composite            STRING,
  composite_desc       STRING,
  laboratory           STRING,
  delivery_method      STRING,
  parent               STRING,
  date_sent            STRING,
  date_received        STRING,
  samp_matrix          STRING,
  samp_reason          STRING,
  samp_technique       STRING,
  sample_class         STRING,
  sample_type          STRING,
  sample_name          STRING,
  static_wl            STRING,
  purge_method         STRING,
  purge_volume         STRING,
  purge_wl             STRING,
  watertype            STRING,
  record_resp          STRING,
  frequency            STRING,
  initial_depth        STRING,
  final_depth          STRING,
  samp_is_composite    STRING,
  samp_colected        STRING,
  not_colected_reason  STRING,
  samp_accreditation   STRING,
  lab_legis            STRING,
  samp_purpose         STRING
);

CREATE TABLE IF NOT EXISTS AGA.bronze.parameter_result (
  station                   STRING,
  sample_id                 STRING,
  chemical_name             STRING,
  result_value              STRING,
  result_unit               STRING,
  qualifier                 STRING,
  reporting_detection_limit STRING,
  sampling_precision        STRING,
  fraction_code             STRING,
  analysis_method           STRING,
  outlier                   STRING,
  cas_number                STRING,
  comment                   STRING,
  analysis_date             STRING,
  dilution_factor           STRING,
  result_txt                STRING,
  reporting_quanti_limit    STRING,
  original_parameter        STRING,
  original_result           STRING,
  original_unit             STRING,
  parameter_group           STRING,
  record_resp               STRING,
  report_type               STRING,
  analysis_type             STRING,
  result_reportable         STRING,
  qaqc_flag                 STRING,
  result_accreditation      STRING
);

CREATE TABLE IF NOT EXISTS AGA.bronze.gw_level (
  station          STRING,
  screen_id        STRING,
  date_            STRING,
  depth_           STRING,
  dry_indicator    STRING,
  comment          STRING,
  data_source      STRING,
  data_type        STRING,
  instrument_type  STRING,
  level_type       STRING,
  level_elevation  STRING,
  equipe_medicao   STRING,
  pressure         STRING,
  record_resp      STRING,
  status           STRING,
  data_aga         STRING,
  outlier          STRING,
  dry_depth_m      STRING,
  id               STRING
);

CREATE TABLE IF NOT EXISTS AGA.bronze.flow_rate (
  id              STRING,
  station         STRING,
  date_           STRING,
  flow_rate       STRING,
  dry_indicator   STRING,
  status          STRING,
  data_type       STRING,
  instrument_type STRING,
  frequency       STRING,
  measuring_resp  STRING,
  comment         STRING,
  data_source     STRING,
  qaqc_flag       STRING,
  inserted_by     STRING,
  inserted_at     STRING,
  updated_at      STRING,
  updated_by      STRING,
  original_value  STRING,
  original_unit   STRING,
  sample_id       STRING
);

-- =====================================================================
-- SILVER (tipado, chaves conforme diagrama)
-- =====================================================================

CREATE TABLE IF NOT EXISTS AGA.silver.station (
  id                INT NOT NULL,
  name              STRING NOT NULL,
  x                 DOUBLE,
  y                 DOUBLE,
  elevation         DOUBLE,
  toc               DOUBLE,
  stn_type          STRING,
  depth             DOUBLE,
  purpose           STRING,
  comment           STRING,
  construction_date DATE,
  update_date       DATE,
  data_source       STRING,
  demand_type       STRING,
  stn_description   STRING,
  matrix            STRING,
  bussiness_unit    STRING,
  photo             STRING,
  record_resp       STRING,
  status            STRING,
  watershed         STRING,
  frame_legislation STRING,
  water_body        STRING,
  current_elevation DOUBLE,
  sub_watershed     STRING,
  stn_class         STRING,
  paebm             STRING,
  mecanica_rochas   STRING,
  geotecnia         STRING,
  meio_ambiente     STRING,
  comment_name      STRING,
  start_monitoring  DATE,
  end_monitoring    DATE,
  CONSTRAINT pk_station PRIMARY KEY (id)
);

CREATE TABLE IF NOT EXISTS AGA.silver.project (
    ID INT,
    Name STRING,
    plsd STRING,
    Client INT,
    PDA INT,
    List INT,
    Soil INT,
    Description STRING,
    Path STRING,
    method_name STRING,
    chemical_name STRING,
    CONSTRAINT pk_project PRIMARY KEY (ID)
);

CREATE TABLE IF NOT EXISTS AGA.silver.pstation (
    ID INT,
    SID INT,
    x STRING,
    y STRING,
    prj_sys INT,
    CONSTRAINT pk_pstation PRIMARY KEY (ID, SID)
);

CREATE TABLE IF NOT EXISTS AGA.silver.screen (
  station         INT NOT NULL,
  screen_id       INT NOT NULL,
  from_           DOUBLE,
  to_             DOUBLE,
  diam_           DOUBLE,
  screen_type     STRING,
  slot_number     INT,
  screen_material STRING,
  comment         STRING,
  data_source     STRING,
  record_resp     STRING,
  CONSTRAINT pk_screen PRIMARY KEY (station, screen_id)
);

CREATE TABLE IF NOT EXISTS AGA.silver.parameter_sample (
  station              INT NOT NULL,
  sample_id            STRING NOT NULL,
  sample_date          TIMESTAMP,
  lab_id               STRING,
  batch_id             STRING,
  quality_code         STRING,
  chain_cust           STRING,
  collection_quarter   STRING,
  comment              STRING,
  composite            STRING,
  composite_desc       STRING,
  laboratory           STRING,
  delivery_method      STRING,
  parent               STRING,
  date_sent            TIMESTAMP,
  date_received        TIMESTAMP,
  samp_matrix          STRING,
  samp_reason          STRING,
  samp_technique       STRING,
  sample_class         STRING,
  sample_type          STRING,
  sample_name          STRING,
  static_wl            DOUBLE,
  purge_method         STRING,
  purge_volume         DOUBLE,
  purge_wl             DOUBLE,
  watertype            STRING,
  record_resp          STRING,
  frequency            STRING,
  initial_depth        DOUBLE,
  final_depth          DOUBLE,
  samp_is_composite    BOOLEAN,
  samp_colected        BOOLEAN,
  not_colected_reason  STRING,
  samp_accreditation   STRING,
  lab_legis            STRING,
  samp_purpose         STRING,
  CONSTRAINT pk_parameter_sample PRIMARY KEY (station, sample_id)
);

CREATE TABLE IF NOT EXISTS AGA.silver.parameter_result (
  station                   INT NOT NULL,
  sample_id                 STRING NOT NULL,
  chemical_name             STRING NOT NULL,
  result_value              DOUBLE,
  result_unit               STRING,
  qualifier                 STRING,
  reporting_detection_limit DOUBLE,
  sampling_precision        DOUBLE,
  fraction_code             STRING,
  analysis_method           STRING,
  outlier                   BOOLEAN,
  cas_number                STRING,
  comment                   STRING,
  analysis_date             TIMESTAMP,
  dilution_factor           STRING,
  result_txt                STRING,
  reporting_quanti_limit    DOUBLE,
  original_parameter        STRING,
  original_result           STRING,
  original_unit             STRING,
  parameter_group           STRING,
  record_resp               STRING,
  report_type               STRING,
  analysis_type             STRING NOT NULL,
  result_reportable         BOOLEAN,
  qaqc_flag                 STRING,
  result_accreditation      STRING,
  CONSTRAINT pk_parameter_result PRIMARY KEY (station, sample_id, chemical_name, analysis_type)
);

CREATE TABLE IF NOT EXISTS AGA.silver.gw_level (
  station          INT NOT NULL,
  screen_id        INT NOT NULL,
  date_            TIMESTAMP NOT NULL,
  depth_           DOUBLE,
  dry_indicator    STRING,
  comment          STRING,
  data_source      STRING,
  data_type        STRING,
  instrument_type  STRING,
  level_type       STRING,
  level_elevation  DOUBLE,
  equipe_medicao   STRING NOT NULL,
  pressure         DOUBLE,
  record_resp      STRING,
  status           STRING,
  data_aga         DATE,
  outlier          BOOLEAN,
  dry_depth_m      DOUBLE,
  id               INT NOT NULL,
  CONSTRAINT pk_gw_level PRIMARY KEY (station, screen_id, date_)
);

CREATE TABLE IF NOT EXISTS AGA.silver.flow_rate (
  id              INT NOT NULL,
  station         INT NOT NULL,
  date_           TIMESTAMP NOT NULL,
  flow_rate       DOUBLE,
  dry_indicator   BOOLEAN,
  status          STRING,
  data_type       STRING,
  instrument_type STRING,
  frequency       STRING,
  measuring_resp  STRING,
  comment         STRING,
  data_source     STRING,
  qaqc_flag       STRING,
  inserted_by     STRING,
  inserted_at     TIMESTAMP,
  updated_at      TIMESTAMP,
  updated_by      STRING,
  original_value  STRING,
  original_unit   STRING,
  sample_id       STRING,
  CONSTRAINT pk_flow_rate PRIMARY KEY (station, date_)
);
```
